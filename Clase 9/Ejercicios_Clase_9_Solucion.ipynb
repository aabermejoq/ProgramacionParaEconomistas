# Parte 2 — Crear variables nuevas

Igual que en clase, no modelamos con las variables crudas: primero construimos
variables que capturen mejor la información.

Las herramientas son **las mismas tres** de la clase:

| Herramienta | Qué hace | Ejemplo de la clase |
|---|---|---|
| `.apply(lambda x: ...)` | aplica una función a cada valor de la columna | limpiar `EDUCATION` |
| `np.log1p(x)` | `log(x + 1)`, comprime distribuciones sesgadas | `log_limite` |
| `np.where(cond, a, b)` | el `ifelse()` de R | `ratio_utilizacion` |

```python
df = X.copy()
df['ingreso_alto'] = y.values
```

**Ejercicio 2 — Solución**

Si resolviste `sexo_binario` y `casado` con `np.where()` o con `(df['sex'] == 'Male').astype(int)`,
también está bien: las tres formas son correctas. La consigna pedía `.apply()` solo para que
practicaras la lambda.

```python
df['sexo_binario']   = df['sex'].apply(lambda s: 1 if s == 'Male' else 0)
df['casado']         = df['marital-status'].apply(lambda s: 1 if s == 'Married-civ-spouse' else 0)
df['log_ganancia']   = np.log1p(df['capital-gain'])
df['log_perdida']    = np.log1p(df['capital-loss'])
df['ganancia_neta']  = df['capital-gain'] - df['capital-loss']
df['tiene_ganancia'] = np.where(df['capital-gain'] > 0, 1, 0)

nuevas = ['sexo_binario', 'casado', 'log_ganancia', 'log_perdida',
          'ganancia_neta', 'tiene_ganancia']
df[nuevas].describe().round(3)
```

Ahora definimos la lista de variables del modelo.

Fíjate que metimos `fnlwgt` a propósito. Es el *final weight* de la encuesta: un peso
muestral que el Census usa para expandir la muestra a la población. **No tiene nada
que ver con el ingreso de la persona.** Está ahí como control — a ver si los modelos
se dan cuenta solos de que no sirve.

```python
FEATURES = [
    # variables originales numéricas
    'age', 'education-num', 'hours-per-week', 'fnlwgt',
    # variables nuevas
    'sexo_binario', 'casado', 'log_ganancia', 'log_perdida',
    'ganancia_neta', 'tiene_ganancia'
]

X_model = df[FEATURES]
y_model = df['ingreso_alto']

print(f"{len(FEATURES)} variables, {len(X_model)} observaciones")
```

---
# Parte 3 — Train/test split y escalado

Copia la receta de la clase. Recuerda los dos detalles que explicamos:

1. `stratify=y_model` → mantiene el mismo % de "ingreso alto" en train y en test.
2. El scaler se ajusta **solo con train** (`.fit_transform`) y a test solo se le
   aplica (`.transform`). Ajustarlo con test sería filtrar información.

**Ejercicio 3 — Solución**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_model, y_model,
    test_size    = 0.2,
    stratify     = y_model,
    random_state = 42
)

print(f"Train: {X_train.shape[0]} observaciones")
print(f"Test:  {X_test.shape[0]} observaciones")
print(f"\n% de ingreso alto en train: {y_train.mean()*100:.1f}%")
print(f"% de ingreso alto en test:  {y_test.mean()*100:.1f}%")
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)

# verificación: la primera columna es 'age'
print(f"Media de age en train (debe ser ~0): {X_train_scaled[:, 0].mean():.4f}")
print(f"Std de age en train (debe ser ~1):   {X_train_scaled[:, 0].std():.4f}")
```

# Parte 4 — Lasso

**✍️ Pregunta 2 — Solución**

Si contestaste `fnlwgt`, tu razonamiento era impecable: es el peso muestral, no debería
predecir nada. Y **casi** pasa — su coeficiente queda en 0.0377, el segundo más chico de
los que sobreviven. Pero el que efectivamente se va a cero es **`log_ganancia`**.

Esta sorpresa es todo el punto del ejercicio. `log_ganancia`, `ganancia_neta` y
`tiene_ganancia` son **tres versiones de la misma información** (la ganancia de capital).
Están altamente correlacionadas. Y aquí está la idea que hay que llevarse:

> Lasso no elimina variables **inútiles**. Elimina variables **redundantes**.

Del grupo de tres se queda con la que mejor explica (`ganancia_neta`, coeficiente 1.90,
el más grande de todos) y tira el resto. Que una variable tenga coeficiente cero **no**
significa que no sirva — puede significar que otra ya está diciendo lo mismo.

Sobre el coeficiente más grande: `ganancia_neta` (1.90), seguido de `casado` (1.14).
Que `casado` sea el segundo sorprende, y conviene detenerse ahí un segundo: **no** significa
que casarse te suba el sueldo. En la encuesta de 1994, el estatus "Married-civ-spouse" viene
correlacionado con edad, género y participación laboral. Es **correlación, no causalidad** —
exactamente la distinción entre econometría y ML con la que abrimos la clase 9. Este modelo
predice bien; no explica nada.

---

**Ejercicio 4 — Solución**

```python
lasso = LogisticRegression(
    penalty      = 'l1',
    C            = 0.01,
    solver       = 'liblinear',
    max_iter     = 1000,
    random_state = 42
)

lasso.fit(X_train_scaled, y_train)

coeficientes = pd.Series(
    lasso.coef_[0],
    index=FEATURES
).sort_values(key=abs, ascending=False)

print("Coeficientes Lasso (0 = variable eliminada):")
print(coeficientes.round(4))
print(f"\nVariables eliminadas: {(coeficientes == 0).sum()} de {len(FEATURES)}")
```

**Ejercicio 5.** Grafica los coeficientes. Cópiale la gráfica a la clase 9.

```python
coef_no_cero = coeficientes[coeficientes != 0].sort_values()

fig, ax = plt.subplots(figsize=(8, 5))
colores = ['firebrick' if v > 0 else 'steelblue' for v in coef_no_cero.values]

coef_no_cero.plot(kind='barh', ax=ax, color=colores, alpha=0.85)

ax.axvline(0, color='black', linewidth=0.8)
ax.set_title('Coeficientes Lasso (rojo = sube P(>50K), azul = la baja)', fontsize=12)
ax.set_xlabel('Coeficiente')
plt.tight_layout()
plt.show()

```

# Parte 5 — Evaluar: la trampa del accuracy

Ya sabes cómo termina esta historia: el dataset está desbalanceado, así que **el accuracy
solo, miente**. Vamos a demostrarlo otra vez con estos datos.

**Ejercicio 6 — Solución**

```python
y_pred_lasso = lasso.predict(X_test_scaled)
y_prob_lasso = lasso.predict_proba(X_test_scaled)[:, 1]

cm_lasso = confusion_matrix(y_test, y_pred_lasso)

fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay(
    confusion_matrix = cm_lasso,
    display_labels   = ['<=50K', '>50K']
).plot(ax=ax, colorbar=False, cmap='Blues')
ax.set_title('Matriz de confusión — Lasso', fontsize=13)
plt.tight_layout()
plt.show()

tn, fp, fn, tp = cm_lasso.ravel()   # ravel() aplana la matriz 2x2 en 4 números

print(f"Verdaderos Negativos (TN): {tn:>5}")
print(f"Falsos Positivos     (FP): {fp:>5}")
print(f"Falsos Negativos     (FN): {fn:>5}")
print(f"Verdaderos Positivos (TP): {tp:>5}")
print(f"\nDe cada 100 personas que SÍ ganan >50K, el modelo detectó {tp/(tp+fn)*100:.0f}")
```

```python
print("Reporte de clasificación — Lasso:")
print(classification_report(y_test, y_pred_lasso, target_names=['<=50K', '>50K']))

auc_lasso = roc_auc_score(y_test, y_prob_lasso)
print(f"AUC-ROC: {auc_lasso:.4f}")
```

**✍️ Pregunta 3 — Solución**

1. Lasso tiene 84.2%, el modelo tonto 76.1%. La diferencia es de solo **8 puntos**.
   Después de todo el trabajo — construir variables, escalar, entrenar, regularizar —
   ganamos 8 puntos sobre una regla de una línea que no mira absolutamente nada.
   Visto solo con accuracy, el esfuerzo se ve ridículo.

2. Sirvió muchísimo, y el accuracy no lo muestra. El modelo tonto tiene **F1 = 0**:
   detecta **cero** personas de ingreso alto. Es 76% "correcto" y 100% inútil.
   El Lasso tiene **F1 = 0.627** e identifica correctamente a 1,299 personas de ingreso alto.

   La métrica que los distingue es el **F1** (o el recall). ¿Por qué? Porque el accuracy
   promedia el desempeño sobre las dos clases, y como una clase es el 76% de los datos,
   ese promedio queda dominado por ella. El F1 de la clase ">50K" solo mira la clase que
   nos importa, y ahí el modelo tonto no tiene dónde esconderse.

   **La moraleja general:** cuando las clases están desbalanceadas, el accuracy premia
   a los modelos que ignoran la clase rara — que es justo la clase que uno quiere predecir.
   Si a alguien le presumen un modelo con "90% de accuracy" sin decirle la tasa base,
   no le dijeron nada.

# Parte 6 — Overfitting

Esta es **la gráfica más importante de la clase 9**. Vamos a reproducirla aquí.

La idea: entrenar árboles cada vez más profundos y medir el AUC en train y en test
**al mismo tiempo**.

**✍️ Pregunta 4 — Solución**

1. **Train solo sube**, siempre, de forma monótona. Llega a **0.992** en profundidad 25.
   Un árbol más profundo *nunca* puede memorizar peor el training: en el peor caso hace
   los mismos cortes de antes y agrega más. Por eso una curva de train que sube no es
   evidencia de nada.

2. **Test sube, toca techo y baja.** Pasa de 0.765 (depth 1) a 0.906 (depth 8) y de ahí
   se desploma hasta 0.799 (depth 25). A partir del punto dulce, cada nivel extra de
   profundidad **empeora** el modelo en datos nuevos.

3. El máximo está en **profundidad 8** (AUC test = 0.906). Si contestaste entre 5 y 12,
   tu intuición va bien. Si contestaste "25" o "entre más profundo, mejor", regresa a la
   gráfica del Paso 8 de la clase antes de seguir: es la idea central del curso y esta
   tarea entera se apoya en ella.

Y el número que resume todo: en profundidad 25 el árbol tiene **0.992 de AUC en train
y 0.799 en test**. Esa brecha de casi 20 puntos no es un bug ni mala suerte — es el árbol
memorizando el ruido de las 39,073 personas del training. Cerrar esa brecha es
literalmente de lo que se trata el machine learning aplicado.

---

**Ejercicio 8 — Solución**

```python
profundidades = range(1, 26)
auc_train = []
auc_test  = []

for d in profundidades:
    arbol = DecisionTreeClassifier(max_depth=d, random_state=42)
    arbol.fit(X_train, y_train)

    auc_train.append(roc_auc_score(y_train, arbol.predict_proba(X_train)[:, 1]))
    auc_test.append(roc_auc_score(y_test,  arbol.predict_proba(X_test)[:, 1]))

mejor = list(profundidades)[int(np.argmax(auc_test))]

fig, ax = plt.subplots(figsize=(11, 6))
ax.plot(list(profundidades), auc_train, 'o-', color='firebrick', lw=2,
        label='TRAIN (datos que ya vio)')
ax.plot(list(profundidades), auc_test, 'o-', color='steelblue', lw=2,
        label='TEST (datos nuevos — lo que importa)')
ax.axvline(mejor, color='green', ls='--', alpha=0.7)
ax.axvspan(mejor, 25, alpha=0.08, color='red')
ax.set_xlabel('Profundidad del árbol (más a la derecha = más complejo)', fontsize=12)
ax.set_ylabel('AUC (1 = perfecto, 0.5 = azar)', fontsize=12)
ax.set_title('El overfitting en una sola gráfica', fontsize=15)
ax.legend(fontsize=11, loc='lower right')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Mejor profundidad según TEST: {mejor}")
print(f"AUC máximo en TRAIN: {max(auc_train):.3f}")
print(f"AUC máximo en TEST:  {max(auc_test):.3f}")
print(f"En profundidad 25 → train: {auc_train[-1]:.3f} | test: {auc_test[-1]:.3f}")
```

# Parte 7 — Random Forest

Un árbol solo tiene alta varianza. La solución: entrenar 200 árboles distintos y
promediar sus votos (**bagging**).

**Ejercicio 9 — Solución**

```python
rf = RandomForestClassifier(
    n_estimators     = 200,
    max_depth        = 15,
    min_samples_leaf = 10,
    n_jobs           = -1,
    random_state     = 42
)

# los árboles NO necesitan escalado → X_train, no X_train_scaled
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)
y_prob_rf = rf.predict_proba(X_test)[:, 1]

fig, ax = plt.subplots(figsize=(5.5, 4.5))
ConfusionMatrixDisplay(
    confusion_matrix = confusion_matrix(y_test, y_pred_rf),
    display_labels   = ['<=50K', '>50K']
).plot(ax=ax, colorbar=False, cmap='Greens')
ax.set_title('Matriz de confusión — Random Forest', fontsize=13)
plt.tight_layout()
plt.show()

f1_rf  = f1_score(y_test, y_pred_rf)
auc_rf = roc_auc_score(y_test, y_prob_rf)
print(f"F1 (>50K) — Random Forest: {f1_rf:.3f}")
print(f"AUC-ROC   — Random Forest: {auc_rf:.4f}")
```

**✍️ Pregunta 5 — Solución**

**Sobre el ranking:** parecido en el espíritu, distinto en el detalle. Comparación:

| | Lasso (|coef|) | Random Forest (importancia) |
|---|---|---|
| 1º | ganancia_neta (1.90) | casado (0.298) |
| 2º | casado (1.14) | education-num (0.184) |
| 3º | education-num (0.89) | ganancia_neta (0.157) |
| 4º | hours-per-week (0.36) | **log_ganancia (0.122)** |

El contraste importante: `log_ganancia` es la variable que Lasso **eliminó**
(coef = 0) y el bosque la pone en **cuarto lugar**. No se contradicen — están midiendo
cosas distintas. Lasso, ante variables redundantes, escoge una y tira el resto. El bosque
no tiene esa presión: los 200 árboles reparten el crédito entre las tres versiones de la
ganancia de capital, y cada una recibe su parte.

**La lección:** "importancia" no es una propiedad de la variable, es una
propiedad **del modelo**. Dos modelos igual de buenos pueden contar historias distintas
sobre qué variable importa. Por eso el ranking de importancias no sirve para inferencia
causal — sirve para entender qué está usando *este* modelo.

**Sobre `fnlwgt`:** aquí está lo interesante. No queda en cero: recibe **0.0282**, más que
`sexo_binario` (0.0150). Una variable que **sabemos con certeza** que no tiene ninguna
relación con el ingreso obtiene importancia positiva y le gana a una que sí importa.

¿Por qué pasa? `fnlwgt` es continua y tiene miles de valores distintos, así que el árbol
siempre encuentra *algún* corte que mejora un poquito el gini — está ajustando ruido.
`sexo_binario` solo tiene dos valores: solo hay un corte posible, y una sola oportunidad
de acumular importancia. **Las importancias del Random Forest están sesgadas a favor de
las variables con muchos valores distintos.** Es un sesgo conocido y documentado del método.

Si contestaste "va a quedar en cero", tenías toda la lógica de tu lado y el output te
desmiente. Ese era el ejercicio: `fnlwgt` estaba ahí desde la Parte 2 como trampa.
La conclusión práctica: **una gráfica de importancias no se lee como una tabla de
regresión.** Que una variable aparezca arriba no prueba que importe, y que aparezca
abajo no prueba que no.

# Parte 8 — Los dos modelos, lado a lado

**Ejercicio 11 — Solución**

```python
fig, ax = plt.subplots(figsize=(7.5, 7))

modelos_roc = [
    ('Lasso',         y_prob_lasso, 'steelblue'),
    ('Random Forest', y_prob_rf,    'seagreen'),
]

for nombre, prob, color in modelos_roc:
    fpr, tpr, _ = roc_curve(y_test, prob)
    auc = roc_auc_score(y_test, prob)
    ax.plot(fpr, tpr, color=color, lw=2.5, label=f'{nombre} (AUC = {auc:.3f})')

ax.plot([0, 1], [0, 1], 'k--', alpha=0.5, label='Azar (AUC = 0.5)')
ax.set_xlabel('Falsas alarmas (FPR)', fontsize=12)
ax.set_ylabel('Ingresos altos detectados (Recall)', fontsize=12)
ax.set_title('Curva ROC — Lasso vs Random Forest', fontsize=13)
ax.legend(fontsize=11, loc='lower right')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
